## Task 1: Data imputation

```python
import pandas as pd, numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('imu_walk.csv')
print(df.head(10))

# Interpolate missing values (if any) and forward-fill or backward-fill leftovers
#df_interp = df.interpolate(method='linear')
df_interp = df.interpolate(method='linear').ffill().bfill()
df_interp.head()
```

## Task 2: Low-variance and redundant feature removal

```python
from sklearn.feature_selection import VarianceThreshold

num_cols = df_interp.select_dtypes(include=[float,int]).columns.tolist()

vt = VarianceThreshold(threshold=1e-5)  # tight threshold to catch near-constant features
vt.fit(df_interp[num_cols])
kept_mask = vt.get_support()  # boolean, true for the columns to keep, false to remove

removed_low_var = [c for c, keep in zip(num_cols, kept_mask) if not keep]  # zip creates an iterator of (column, keep_flag) pairs
#col_tuple = zip(num_cols, kept_mask) 
#print(list(col_tuple))

print('Removed low-variance:', removed_low_var)
pd.DataFrame({'variance': df_interp[num_cols].var()}).sort_values('variance').head(10)
```

### Redundant feature detection (correlations)

The example below illustrates one way to handle redundancy, but it is **NOT** a definitive solution. What you want to achieve from your dataset during preprocessing should always guide your approach.

```python
import numpy as np

corr = df_interp[num_cols].corr().abs()  # corr: compute pairwise Pearson correlation between columns
upper = corr.where(np.triu(np.ones(corr.shape), k=1).astype(bool))
#print(np.triu(np.ones(corr.shape), k=1))
#print(np.triu(np.ones(corr.shape), k=1).astype(bool))
#print(upper)
      
high_corr_pairs = [(c1, c2, upper.loc[c1, c2]) 
                   for c1 in upper.index for c2 in upper.columns 
                   if pd.notna(upper.loc[c1, c2]) and upper.loc[c1, c2] > 0.95]  # notna: check that the value is not NaN (i.e., valid/non-missing)

print('High-corr pairs (>0.95):', high_corr_pairs[:10])
```

```python
# Greedy drop: remove the second feature in each high-correlation pair

to_drop = set()  # set: create a set of unique elements (no duplicates)

for c1, c2, v in high_corr_pairs:
    if c2 not in to_drop and c1 not in to_drop:
        to_drop.add(c2)

selected = [c for c in num_cols if c not in removed_low_var and c not in to_drop]

print(to_drop)
print(selected)
```

## Task 3: Standardization and visualization

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X = df_interp[selected].values
X_std = scaler.fit_transform(X)
df_std = pd.DataFrame(X_std, index=df_interp.index, columns=selected)

# visualizing distribution shifts (histograms) and time-series overlays 
cols_to_plot = [c for c in ['ax', 'ay', 'az'] if c in selected][:3]

fig, axes = plt.subplots(2, len(cols_to_plot), figsize=(5 * len(cols_to_plot), 8))

# axes[0, :] → histogram row / axes[1, :] → time-series row
for i, c in enumerate(cols_to_plot):
    # --- (1) Histogram ---
    axes[0, i].hist(df_interp[c].dropna().values, bins=40, alpha=0.6, label=f'{c} (raw)')  
    # bins=40: divide the data range into 40 intervals (bars) to show the distribution
    axes[0, i].hist(df_std[c].dropna().values, bins=40, alpha=0.6, label=f'{c} (standardized)')
    axes[0, i].set_title(f'Distribution — {c}')
    axes[0, i].legend()

    # --- (2) Time-series ---
    axes[1, i].plot(df_interp.index, df_interp[c], label=f'{c} raw')
    axes[1, i].plot(df_std.index, df_std[c], label=f'{c} standardized', alpha=0.8)
    axes[1, i].set_title(f'Time series — {c}')
    axes[1, i].legend()

plt.suptitle('Before vs After Standardization (Histograms & Time Series)', fontsize=14, y=1.02)
plt.tight_layout()
plt.show()
```

## Task 4: Filtering the signal

```python
from scipy.signal import butter, sosfiltfilt

def lowpass(x, cutoff, fs=50, order=4):
    sos = butter(order, cutoff, btype='low', fs=fs, output='sos')
    return sosfiltfilt(sos, x)

def highpass(x, cutoff, fs=50, order=4):
    sos = butter(order, cutoff, btype='high', fs=fs, output='sos')
    return sosfiltfilt(sos, x)

def bandpass(x, low, high, fs=50, order=4):
    sos = butter(order, [low, high], btype='bandpass', fs=fs, output='sos')
    return sosfiltfilt(sos, x)
    
    # filtfilt() applies the filter forward and backward to the signal,
    # effectively doubling the filter order but removing phase distortion.
    # This ensures the filtered signal has no time delay (zero phase shift).
    # sos: Second-Order Sections, numerically stable filter representation for high-order filters
```

```python
ax = df_std['ax'].astype(float).values
ay = df_std['ay'].astype(float).values
az = df_std['az'].astype(float).values
t = df['t'].astype(float).values

# Magnitude
acc_mag = np.sqrt(ax**2 + ay**2 + az**2)

# Low-pass filtering on magnitude for simple visual comparison
mag_lp5  = lowpass(acc_mag, 5)
mag_lp10 = lowpass(acc_mag, 10)
mag_hp   = highpass(acc_mag, 0.25)
mag_bp   = bandpass(acc_mag, 0.5, 6)

plt.figure(figsize=(10,5))
plt.plot(t, acc_mag, label="Raw", color='gray', alpha=0.3, linewidth=1.0)
plt.plot(t, mag_lp5,  label=f"Low-pass 5Hz")
plt.plot(t, mag_lp10, label=f"Low-pass 10Hz")
plt.plot(t, mag_hp, label=f"High-pass 0.25Hz")
plt.plot(t, mag_bp, label=f"band-pass between 0.5 and 6Hz")
plt.xlabel("Time (s)")
plt.ylabel("Acceleration magnitude")
plt.title("Raw vs filtered magnitude")
plt.legend()
plt.tight_layout()
plt.show()
```

## Task 5: Feature engineering
The following is an example of feature engineering, but it should not be considered the only solution. Use your engineering judgment and domain knowledge to identify meaningful features.

```python
from scipy.signal import find_peaks

env = np.abs(mag_bp)

# calculate adaptive thresholds for finding peaks
med = np.median(env)
mad = np.median(np.abs(env - med)) + 1e-9  # median Absolute Deviation
k_mad = 1.25  # tuning parameter
thr = med + k_mad * mad

fs = 50
refrac_sec = 0.2  # minimum peak distance, refractory period
peaks, props = find_peaks(env, height=thr, distance=int(fs * refrac_sec))

print(f"Number of detected peaks: {len(peaks)}")

plt.figure(figsize=(10, 4))
plt.plot(t, env, label="Envelope (abs)", linewidth=1)
plt.plot(t[peaks], env[peaks], 'go', label='Detected Peaks')
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.title("Envelope-based Peak Detection")
plt.legend()
plt.tight_layout()
plt.show()
```
