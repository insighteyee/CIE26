```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
from sklearn.ensemble import RandomForestRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.svm import SVR
from sklearn.metrics import mean_squared_error,r2_score
import numpy as np
import pandas as pd
```

```python
dataset = pd.read_csv('MELBOURNE_HOUSE_PRICES_LESS.csv',delimiter=',')
dataset.head()
```

```python
dataset = dataset.drop(columns=['Suburb','Address','CouncilArea','Method'])
```

```python
dataset.dropna(inplace=True)
```

```python
dataset.shape
```

```python
#inplace=True ; replaces the same dataset without creating a new dataset
```

```python
dataset = pd.get_dummies(dataset, drop_first=True)
```

```python
#drop_first=True ;  doesn't make extra columns, reduces the number of columns by combining the information
#Ex:     
#			    Furnished 	 semi furnished	
#Furnished 		    1			0	
#semi-furnished	    0			1	
#Unfurnished		0			0	
```

```python
dataset.head()
```

```python
dataset['Price'].shape
```

```python
#minmax normalization between 0 and 1;
#norm = x- xmin/xmax-xmin;
#Reshaping because rank 2 array is needed for the fit_transform command 
```

```python
scalerx = MinMaxScaler()
scalery = MinMaxScaler()

x = scalerx.fit_transform(dataset.drop(['Price'],axis=1))
y = scalery.fit_transform(np.reshape(dataset['Price'],(len(dataset),1)))
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,np.squeeze(y),test_size=0.2,random_state=27)
```

The C parameter in SVM allows us to control the trade-off between the margin and misclassifications. A smaller value of C promotes a wider margin but allows more misclassifications, while a larger value of C prioritizes correct classification at the expense of a narrower margin.

Epsilon in SVM stands for epsilon insensitive loss. It is a loss function that is used in support vector regression (SVR).

```python
# IMPORTANT - ASKED IN FINAL PRESENTATION
RF = RandomForestRegressor(n_estimators=100, max_depth=100)
SVM = SVR(C=0.1,epsilon=0.2)
DT = DecisionTreeRegressor(max_depth=100)
```

```python
# Random Forest training

RF.fit(x_train,y_train)

RFpredict = RF.predict(x_test)
```

```python
# Decision tree training

DT.fit(x_train,y_train)

DTpredict = DT.predict(x_test)
```

```python
# SVM for regression (SVR)

SVM.fit(x_train,y_train)

SVMpredict = SVM.predict(x_test)
```

```python
# the closer to zero the better
print(f"MSE for Random Forest:  {mean_squared_error(y_test, RFpredict)}")  # best one
print(f"MSE for Decision Trees: {mean_squared_error(y_test, DTpredict)}")
print(f"MSE for SVM:            {mean_squared_error(y_test, SVMpredict)}")

# what if random forrest was just lucky or somthing...
# 
```

R-squared (R2) is defined as a number that tells you how well the independent variable(s) in a statistical model explains the variation in the dependent variable.
Best possible score is 1.0 and it can be negative (because the model can be arbitrarily worse). 

```python
print(f"R2 score for Random Forest:  {r2_score(y_test, RFpredict)}")
print(f"R2 score for Decision Trees: {r2_score(y_test, DTpredict)}")
print(f"R2 score for SVM:            {r2_score(y_test, SVMpredict)}")
```
